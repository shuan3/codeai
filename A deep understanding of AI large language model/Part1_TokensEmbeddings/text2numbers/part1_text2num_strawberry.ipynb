```python
# GPT-4's tokenizer
!pip install tiktoken
import tiktoken
tokenizer = tiktoken.get_encoding('cl100k_base')
```

# Count the r's in "strawberry"

```python
strawberry = tokenizer.encode('rapist')
print(strawberry)
for t in strawberry:
  print(f'{t:5d} = {tokenizer.decode([t])}')

strawberry = tokenizer.encode('strawberry')
print(strawberry)
for t in strawberry:
  print(f'{t:5d} = {tokenizer.decode([t])}')
```

```python
r = tokenizer.encode('r')
print(f'{r} = {tokenizer.decode(r)}')
```

```python
print(strawberry)
r in strawberry
[496, 675, 15717] in strawberry
```

# Conclusion: There are no "r"s in "strawberry!

```python
# decode the encoding to get back to a string for the word
strawberry_string = tokenizer.decode(strawberry)
strawberry_string
```

```python
# and for the letter
r_string = tokenizer.decode(r)
r_string
```

```python
# and now we can get the correct answer!
strawberry_string.count(r_string)
```

```python
from transformers import BertTokenizer
bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
dir(tokenizer)
```

```python
all_tokens=list(bert_tokenizer.get_vocab().keys()  )
print(all_tokens[20000:20100])
all_tokens=list(bert_tokenizer.get_vocab().values()  )
print(all_tokens[20000:20100])
# print(tokenizer.vocab_size)
print(bert_tokenizer.vocab_size)
bert_tokenizer.get_vocab()['science']   
```

```python
word='science'
res1=bert_tokenizer.convert_tokens_to_ids(word)
res2=bert_tokenizer.get_vocab()[word]
print(f'{word} = {tokenizer.encode(word)} in GPT-4 tokenizer')
print(f'{word} = {res2} in BERT tokenizer')
```

```python
word='science is grw=!!t'
res1=bert_tokenizer.convert_tokens_to_ids(word)

print(f'{word} = {res1} in BERT tokenizer')
res3=bert_tokenizer.encode(word)
for i in res3:
    print(f"Token {i} = {bert_tokenizer.decode([i])}")
print(bert_tokenizer.decode(res3,skip_special_tokens=False))
print(bert_tokenizer.decode(res3,skip_special_tokens=True))
```
